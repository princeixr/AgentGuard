# AgentGuard Session Risk Analysis

Use this notebook to inspect guard decisions and session risk records.

```python
import json, os, urllib.request

ELASTICSEARCH_URL = os.environ.get('ELASTICSEARCH_URL', '').rstrip('/')
ELASTICSEARCH_API_KEY = os.environ.get('ELASTICSEARCH_API_KEY', '')
DECISION_INDEX = os.environ.get('AGENTGUARD_ELASTIC_GUARD_DECISIONS_INDEX', 'agentguard-guard-decisions-v1')

headers = {'Content-Type': 'application/json'}
if ELASTICSEARCH_API_KEY:
    headers['Authorization'] = f'ApiKey {ELASTICSEARCH_API_KEY}'
```

```python
query = {
    'size': 25,
    'sort': [{'@timestamp': {'order': 'desc'}}],
    'query': {'match_all': {}},
}
request = urllib.request.Request(
    f'{ELASTICSEARCH_URL}/{DECISION_INDEX}/_search',
    data=json.dumps(query).encode('utf-8'),
    headers=headers,
    method='POST',
)
with urllib.request.urlopen(request) as response:
    result = json.loads(response.read().decode('utf-8'))
[(hit['_source']['session_id'], hit['_source']['step_index'], hit['_source']['decision'], hit['_source']['final_risk_score']) for hit in result['hits']['hits']]
```
